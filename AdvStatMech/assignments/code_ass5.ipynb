```julia
using DelimitedFiles, Plots, LaTeXStrings
pgfplotsx()
```

```julia
A =readdlm("data.dat");
T =(A[2:end,1])
H = [parse(Int, replace(A, "," => "")) for A in A[2:end,2]]
M = A[2:end,3];
Tc =  627.2
t = (T .- Tc)./Tc
tp = findall(x->x>0, t)
tm = findall(x->x<0, t);
beta=0.368 
delta=4.22

plot(H[tp]./(t[tp]).^(beta*delta), M[tp]./(t[tp]).^beta, marker=:circle)

plot()
ind614 = findall(x->x==614.41, T)
plot!(H[ind614]./(-t[ind614]).^(beta*delta), M[ind614]./(-t[ind614]).^beta, marker=:circle, label=L"\mathcal{F}_-")

ind629 = findall(x->x==629.53 || x==653.31, T)
plot!(H[ind629]./(t[ind629]).^(beta*delta), M[ind629]./(t[ind629]).^beta, marker=:circle,label=L"\mathcal{F}_+")


plot!(legend=:bottomright)
xlabel!(L"\frac{h}{|t|^{\beta\delta}}")
ylabel!(L"\frac{m}{|t|^\beta}")
plot!(tickfontsize=18, labelfontsize=26,guidefontsize=16, legendfontsize=15, title=L"T_c= 627.2\ \ \beta=0.368\ \ \delta=4.22")
```
