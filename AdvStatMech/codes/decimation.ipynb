```julia
gr()
iter = 10
K0 = 3.5
K = K0
f(x) = 0.5*log(cosh(2*x))
Kiter=[]
for n in 1:iter 
    K = 0.5*log(cosh(2*K))
    push!(Kiter, K)
end
Kv = LinRange(0,4,50) 
plot(Kv, Kv, ls=:dash, c="black", label="")
plot!(Kv, (@. 0.5*log(cosh(2*Kv))), c="green", label="")
plot!([K0, K0], [0, Kiter[1]],c=cmap[1], lw=2, label="", arrow=true)

plot!([K0,Kiter[1]],[Kiter[1], Kiter[1]],c=cmap[2], lw=2, label="", arrow=true)
c1 = RGB(0, 0, 1)
c2 = RGB(1, 0, 0)
cmap = [RGB( (1 - t) * red(c1) + t * red(c2), (1 - t) * green(c1) + t * green(c2), (1 - t) * blue(c1) + t * blue(c2) ) for t in range(0, 1, length=iter+1)]


for i =1:iter-1
plot!([Kiter[i], Kiter[i]],[Kiter[i], Kiter[i+1]], label="", c=cmap[i+1], lw=2, arrow=true)
plot!([Kiter[i], Kiter[i+1]],[Kiter[i+1], Kiter[i+1]], label="", c=cmap[i+2], lw=2, arrow=true)
end
plot!(xlabel=L"K", size=(500,450), ticksize=20)
```
