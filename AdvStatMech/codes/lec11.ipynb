```julia
using Plots

L(m,p,T) = @. (p[1]/2)*(T-1)*m^2 - (p[2]/3)*m^3 + (p[3]/4)*m^4

m = LinRange(-0.3,1,1000)
p = [1,3,4]

Tvals = [2.0, 1.52, 1.45, 1.375, 1.15, 1.0, 0.2]
Tstr = ["T>T^*", "T=T^{*}", "T^{**} < T < T^{*}", "T=T^{**} ",
        "Tc <T < T^{**}", "T=Tc", "T < Tc"]

# Tvals = [2.0, 1.5]
# Tstr = ["2.0 > T*","T* = 1.5625"]


plot()
for (i,T) in enumerate(Tvals)
    plot!(m, L(m,p,T) .- L(0,p,T),
          label=L"%$(Tstr[i])", lw=2)
end

xlabel!(L"m")
ylabel!(L"\mathcal{L}")
plot!(size=(500,650), legendfontsize=12, guidefontsize=20,tickfontsize=10)
hline!([0], c="black", ls=:dash, lw=0.4, label="", left_margin=5Plots.mm)
```
