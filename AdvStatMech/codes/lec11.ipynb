```julia
using Plots

L(m,p,T) = @. (p[1]/2)*(T-1)*m^2 - (p[2]/3)*m^3 + (p[3]/4)*m^4

m = LinRange(-0.3,1,1000)
p = [1,3,4]

Tvals = [2.0,1.5625, 1.53,1.5, 1.3,1.0,0.5]
Tstr = ["T>T^*", "T=T^{*}", "T^{**} < T < T^{*}", "T=T^{**} ",
         "T_c <T < T^{**}", "T=T_c", "T < T_c"]

# Tvals = [2.0, 1.5]
# Tstr = ["2.0 > T*","T* = 1.5625"]


plot()
for (i,T) in enumerate(Tvals)
    plot!(m, L(m,p,T) .- L(0,p,T),
          label=L"%$(Tstr[i])", lw=2)
end
#
xlabel!(L"m")
ylabel!(L"\mathcal{L}")
ylims!(-0.05,0.1)
plot!(size=(550,650), legendfontsize=10, guidefontsize=20,tickfontsize=10, legend=:topleft, legendcolumn=1)
p1=hline!([0], c="black", ls=:dash, lw=0.4, label="", left_margin=5Plots.mm, title="(b)")
```

```julia
savefig("first_order.pdf")
```

```julia
using Plots

# Define range for b/c
bc = LinRange(-5, 5, 400)

# First-order line: a/c = (3/16)*(b/c)^2  for b<0
ac_first = (2/9) .* bc.^2
ac_spin = (1/4 ).* bc.^2
# Plot
plot()

# Second-order line (a=0 for b>0)
# plot!([0,0], [0,5], lw=3, label="Second order (a=0)")

# First-order line (only for b<0)
mask = bc .> 0
plot!(ac_first[mask], bc[mask], lw=3, label="First order line")
plot!(ac_spin[mask], bc[mask], label="Spinodal line", lw=1.4, c="grey")
vline!([0], label="", c="grey", ls=:dash, lw=1.3)
# Axes labels
ylabel!(L"\frac{a_3}{a_4}")
xlabel!(L"\frac{a_2t}{a_4}")

# Limits similar to textbook
xlims!(-2,6)
ylims!(-0.5,7)

# Mark tricritical point
scatter!([0],[0], markersize=3, label="", c="black", legend=:topright, guidefontsize=20)

p2= plot!(size=(400,400), title="(a)", tickfontsize=12)
```

```julia
using Plots

# Define range for b/c
bc = LinRange(-5, 5, 400)

# First-order line: a/c = (3/16)*(b/c)^2  for b<0
ac_first = (3/16) .* bc.^2
ac_spin = (1/4) .* bc.^2

# Plot
plot()

# Second-order line (a=0 for b>0)
plot!([0,0], [0,5], lw=3, label="Second order line")

# First-order line (only for b<0)
mask = bc .< 0
plot!(ac_first[mask], bc[mask], lw=3, label="First order line")
plot!(ac_spin[mask], bc[mask], lw=1.3, label="Spinodal line", c="black", ls=:dashdot)

# Axes labels
xlabel!(L"\frac{a_2t}{a_6}")
ylabel!(L"\frac{a_4}{a_6}")

# Limits similar to textbook
xlims!(-2,6)
ylims!(-5,5)

# Mark tricritical point
scatter!([0],[0], markersize=4, label="Tricritical point", legend=:topright, c="green")
hline!([0], ls=:dash, lw=1.3, c="grey", label="")
p3= plot!(size=(400,400), label="", guidefontsize=20, tickfontsize=12)
```

```julia
using Plots

L(m,p,T) = @. (p[1]/2)*(T-1)*m^2 + (p[2]/4)*m^4 + (p[3]/6)*m^6

m = LinRange(-1,1,300)
p = [1,-4,6]

Tvals = [2.0, 1.67, 1.6, 1.5, 1.3, 1.0, 0.8]
Tstr = ["T>T^*", "T=T^{*}", "T^{**} < T < T^{*}", "T=T^{**} ",
         "T_c <T < T^{**}", "T=T_c", "T < T_c"]


plot()
for (i,T) in enumerate(Tvals)
    plot!(m, L(m,p,T) .- L(0,p,T),
          label=L"%$(Tstr[i])", lw=2)
end
#
xlabel!(L"m")
ylabel!(L"\mathcal{L}")
# ylims!(-0.01,0.1)
plot!(size=(550,650), legendfontsize=10, guidefontsize=20,tickfontsize=10, legend=:top, legendcolumn=1)
p1=hline!([0], c="black", ls=:dash, lw=0.4, label="", left_margin=5Plots.mm, title="(b)")
```

```julia
plot(p3, p1, size=(800,350), tickfontsize=12)
```
