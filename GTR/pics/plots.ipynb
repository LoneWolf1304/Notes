```julia
using Plots, LaTeXStrings, PythonPlot
```

```julia
t = LinRange(4,20,500)
a = @. 2*sqrt(t-4)
Plots.plot(t,a, xlim=(0,20), xticks=(4,[L"t_o"]),tickfont=font(14), yticks=[], label="", left_margin=4Plots.mm,guidefont=font(14), size=(400,300))
xlabel!(L"t")
vline!([4], ls=:dash, c="red",label="")
p1 = ylabel!(L"a(t)")
t = LinRange(4.2,20,500)
rho = @. 1/(t-4)^2
Plots.plot(t,rho, xlim=(0,20), xticks=(4,[L"t_o"]),tickfont=font(14), yticks=[], label="", left_margin=4Plots.mm,guidefont=font(14), size=(400,300))
xlabel!(L"t")
vline!([4], ls=:dash, c="red",label="")
p2 = ylabel!(L"\rho(t)")
Plots.plot(p1,p2, size=(800, 300), left_margin=7Plots.mm,bottom_margin=6Plots.mm)
```

```julia
Plots.savefig("lec28_1.pdf")
```
